```python
# Day 18: Select Dataset Columns Using Pandas

## Objective
Learn to select specific rows and columns from a DataFrame using Pandas `loc` and `iloc`.

## Tasks
- Select a single column.
- Select multiple columns.
- Select rows using index labels with `loc`.
- Select rows using integer positions with `iloc`.
- Select specific rows and columns together.

## Dataset
Iris Flower Dataset
```

```python
import pandas as pd
```

```python
df = pd.read_csv("iris.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
# Select one column by its name
sepal_length = df["sepal length (cm)"]

print("First five sepal length values:")
display(sepal_length.head())

print("\nData type:", type(sepal_length))
```

```python
selected_columns = df[
    ["sepal length (cm)", "petal length (cm)", "species"]
]

print("Selected columns:")
display(selected_columns.head())

print("Selected DataFrame shape:", selected_columns.shape)
```

```python
print("Rows with index labels 0 to 4:")
display(df.loc[0:4])

print("\nSelected rows and columns using loc:")
display(
    df.loc[0:4, ["sepal length (cm)", "species"]]
)
```

```python
print("First five rows using iloc:")
display(df.iloc[0:5])

print("\nFirst five rows and first two columns:")
display(df.iloc[0:5, 0:2])
```

```python
# First row, first column
print("Value at first row and first column:")
print(df.iloc[0, 0])

# First row, selected columns by name
print("\nFirst row: sepal length and species")
display(df.loc[0, ["sepal length (cm)", "species"]])

# First five rows, specific columns by integer positions
print("\nFirst five rows, columns at positions 0 and 2:")
display(df.iloc[0:5, [0, 2]])
```

```python
print("Using loc — label-based selection:")
display(df.loc[0:2, ["sepal length (cm)", "species"]])

print("\nUsing iloc — position-based selection:")
display(df.iloc[0:3, [0, 4]])
```

```python
import matplotlib.pyplot as plt
```

```python


sample = df.iloc[0:10]

plt.figure(figsize=(10, 5))
plt.bar(
    sample.index.astype(str),
    sample["sepal length (cm)"]
)

plt.title("Sepal Length of First 10 Iris Records")
plt.xlabel("DataFrame Index")
plt.ylabel("Sepal Length (cm)")
plt.tight_layout()

plt.savefig(
    "day18_selected_data.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
